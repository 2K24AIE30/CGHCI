```python
def rgb_to_grayscale(r: int, g: int, b: int) -> int:
    """Convert one RGB pixel to a rounded grayscale luminosity value."""
    luminosity = 0.299 * r + 0.587 * g + 0.114 * b
    return int(round(luminosity))


sample_result = rgb_to_grayscale(180, 120, 60)

print("Sample result:", sample_result)

assert sample_result == 131

assert rgb_to_grayscale(80, 80, 80) == 80

print("Sample 1 passed.")
print("All tests passed!")
```

```python
import numpy as np


def adjust_brightness_contrast(image: np.ndarray, alpha: float, beta: float) -> np.ndarray:
    """
    Apply the formula:
        O = alpha * I + beta

    Then clip values to the range 0-255.
    """
    image_float = image.astype(np.float32)
    adjusted = alpha * image_float + beta

    # Keep values between 0 and 255
    adjusted = np.clip(adjusted, 0, 255)

    return adjusted.astype(np.uint8)


# Sample image
sample_image = np.array(
    [[100, 150],
     [200, 50]],
    dtype=np.uint8
)

# Apply brightness/contrast adjustment
sample_output = adjust_brightness_contrast(
    sample_image,
    alpha=1.5,
    beta=-20
)

# Expected result
expected_output = np.array(
    [[130, 205],
     [255, 55]],
    dtype=np.uint8
)

# Test
np.testing.assert_array_equal(sample_output, expected_output)

print("Sample output:")
print(sample_output)

# Extra test: value must clip at 255
extra = adjust_brightness_contrast(
    np.array([[240]], dtype=np.uint8),
    alpha=1.0,
    beta=30
)

assert extra[0, 0] == 255

print("Sample 2 passed.")
print("All tests passed successfully!")
```

```python
import numpy as np


# Return 255 when pixel >= threshold; otherwise return 0.
def threshold_image(image: np.ndarray, threshold: int) -> np.ndarray:
    """Convert a grayscale image to black and white."""
    return np.where(image >= threshold, 255, 0).astype(np.uint8)


problem_1_input = np.array(
    [[20, 128, 200],
     [100, 150, 250]],
    dtype=np.uint8
)

problem_1_expected = np.array(
    [[0, 255, 255],
     [0, 255, 255]],
    dtype=np.uint8
)


# Test
np.testing.assert_array_equal(
    threshold_image(problem_1_input, 128),
    problem_1_expected
)

print("Output:")
print(threshold_image(problem_1_input, 128))

print("Problem 1 passed!")
```

```python
# Calculate image memory in bytes
def display_memory_bytes(width: int, height: int, bpp: int) -> int:
    """Return image memory in bytes."""
    return width * height * bpp // 8


# Test
expected_bytes = 1920 * 1080 * 24 // 8

assert display_memory_bytes(1920, 1080, 24) == expected_bytes

print("Memory:", display_memory_bytes(1920, 1080, 24), "bytes")
print("Problem 2 passed!")
```

```python
import numpy as np


# Calculate the average of 9 pixels
def mean_filter_3x3(neighborhood: np.ndarray) -> int:
    """Return the rounded average of 9 pixels."""
    return int(round(np.sum(neighborhood) / 9))


problem_3_input = np.array(
    [[10, 20, 10],
     [30, 50, 30],
     [10, 20, 10]],
    dtype=np.uint8
)


# Test
assert mean_filter_3x3(problem_3_input) == 21

print("Mean value:", mean_filter_3x3(problem_3_input))
print("Problem 3 passed!")
```

```python
import numpy as np


# Contrast stretching
def contrast_stretch(
    image: np.ndarray,
    in_min: float,
    in_max: float,
    out_min: float = 0.0,
    out_max: float = 255.0,
) -> np.ndarray:
    """Change image values from one range to another."""

    # Apply contrast stretching formula
    stretched = (image - in_min) * (out_max - out_min) / (in_max - in_min) + out_min

    # Keep values inside output range
    return np.clip(stretched, out_min, out_max)


problem_4_input = np.array([50, 100, 150], dtype=np.float32)


# Test
expected = np.array([0.0, 127.5, 255.0], dtype=np.float32)

np.testing.assert_allclose(
    contrast_stretch(problem_4_input, 50, 150),
    expected
)

print("Output:")
print(contrast_stretch(problem_4_input, 50, 150))

print("Problem 4 passed!")
```

```python
import numpy as np

def sobel_response(block):
    gx_kernel = np.array([
        [-1, 0, 1],
        [-2, 0, 2],
        [-1, 0, 1]
    ])

    gy_kernel = np.array([
        [-1, -2, -1],
        [0, 0, 0],
        [1, 2, 1]
    ])

    gx = np.sum(block * gx_kernel)
    gy = np.sum(block * gy_kernel)

    strength = np.sqrt(gx**2 + gy**2)

    return gx, gy, strength


problem_5_input = np.array([
    [20, 20, 200],
    [20, 20, 200],
    [20, 20, 200]
])

gx, gy, strength = sobel_response(problem_5_input)

print("Gx =", gx)
print("Gy =", gy)
print("Edge Strength =", strength)

assert gx == 720 and gy == 0

print("Problem 5 passed!")
```
